# Day 1 — Fund Master EDA
Exploring `01_fund_master.csv`: unique fund houses, categories, sub-categories, risk grades, and AMFI code structure.

```python
import pandas as pd

fm = pd.read_csv("../data/raw/01_fund_master.csv")
print("shape:", fm.shape)
fm.head()
```

```python
for col in ["fund_house", "category", "sub_category", "risk_category"]:
    print("=" * 50)
    print(f"{col}  (unique = {fm[col].nunique()})")
    print("=" * 50)
    print(fm[col].value_counts())
    print()
```

```python
print("total funds        :", len(fm))
print("unique amfi_codes  :", fm["amfi_code"].nunique())
print("min amfi_code      :", fm["amfi_code"].min())
print("max amfi_code      :", fm["amfi_code"].max())
print("duplicate codes?   :", fm["amfi_code"].duplicated().any())

fm.groupby("fund_house")["amfi_code"].count().sort_values(ascending=False)
```
